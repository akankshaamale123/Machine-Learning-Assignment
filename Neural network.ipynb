```python
df = df.fillna(df.mean(numeric_only=True))
```

```python
plt.figure(figsize=(20,10))  # outlier present in the dataset
sns.boxplot(data=df)
plt.show()
```

```python
X = df.iloc[:, :-1] #60 feature columns
y = df.iloc[:, -1] #last column=label(M or R)
```

```python
#by useing the capping methods we remove the outlier present in the dataset
for col in X:
    q1 = df[col].quantile(0.25)
    q3 = df[col].quantile(0.75)
    IQR = q3-q1
    upper_extreme = q3 + (IQR*1.5)
    lower_extreme = q1 - (IQR*1.5)

    df.loc[df[col] > upper_extreme, col] = upper_extreme
    df.loc[df[col] < lower_extreme, col] = lower_extreme
```

```python
plt.figure(figsize=(20,10))  # outlier remove from the dataset
sns.boxplot(data=df)
plt.show()
```

```python
classes = df.iloc[:, -1].unique()
print("\n Unique Classes Alphabets:", classes)
print("Total Number of Classes:", len(classes))
```

	Execute necessary data preprocessing steps including data normalization, managing missing values.

```python
# normalize the data so all the features are in same scale
from sklearn.preprocessing import StandardScaler, LabelEncoder

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
X_scaled
```

```python
#convert M/R into 1/0
y = y.ravel() # convert in 1D array
lbl = LabelEncoder()
y_encoded = lbl.fit_transform(y)
y_encoded
```

2. Model Implementation

Construct a basic ANN model using your chosen high-level neural network library. Ensure your model includes at least one hidden layer.


```python
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
from keras.optimizers import Adam

model = Sequential([Dense(32, input_shape = (60,), activation = 'relu'), # 1st layer
                   Dense(16, activation = 'relu'), # 2nd layer
                   Dense(1, activation = 'sigmoid')]) #3rd layer it gives binary output
```

```python
#compile model
model.compile(
    optimizer = 'adam', loss = 'binary_crossentropy', metrics = ['accuracy']
)
```

divide the dataset into training and test sets.

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split( X_scaled, y_encoded, test_size=0.2, random_state=42)
```

```python
print(X_scaled.shape)
print(y.shape)
```

Train your model on the training set and then use it to make predictions on the test set.

```python
# training the model
history = model.fit(X_train, y_train, epochs = 50, batch_size = 8, validation_split = 0.2)
```

Adopt a structured approach like grid search or random search for hyperparameter tuning, documenting your methodology thoroughly.

```python
from sklearn.model_selection import ParameterGrid
import tensorflow as tf
param_grid = {
    'neurons': [16, 32],
    'activation': ['relu', 'tanh'],
    'learning_rate': [0.001, 0.01]
}

grid = ParameterGrid(param_grid)
results = []

for params in grid:
    print("\nTesting params:", params)

    model = Sequential([
        Dense(params['neurons'], activation=params['activation'], input_shape=(60,)),
        Dense(params['neurons'] // 2, activation=params['activation']),
        Dense(1, activation='sigmoid')
    ])

    model.compile(
        optimizer=tf.keras.optimizers.Adam(learning_rate=params['learning_rate']),
        loss='binary_crossentropy',
        metrics=['accuracy']
    )

    history = model.fit( X_train, y_train,epochs=30, batch_size=8,validation_split=0.2)
```

4. Evaluation

●	Employ suitable metrics such as accuracy, precision, recall, and F1-score to evaluate your model's performance.


```python
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, classification_report

y_pred = (model.predict(X_test) > 0.5).astype(int)

accuracy = accuracy_score(y_test,y_pred)
precision = precision_score(y_test, y_pred)
recall = recall_score(y_test, y_pred)

print(f"Accuracy  : {accuracy:.4f}")
print(f"Precision : {precision:.4f}")
print(f"Recall    : {recall:.4f}")
```

```python
print("\nClassification Report:")
print(classification_report(y_test, y_pred))
```
