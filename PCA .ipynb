Task 2: Dimensionality Reduction with PCA:
1.	Standardize the features to ensure they have a mean of 0 and a standard deviation of Implement PCA to reduce the dimensionality of the dataset.


```python
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA

X = df.drop("Type" , axis=1)
y = df['Type']

scalar = StandardScaler()

X_scaled = scalar.fit_transform(X)

X_scaled_df = pd.DataFrame(X_scaled, columns=X.columns)
X_scaled_df

```

```python
pca = PCA()
X_pca = pca.fit_transform(X_scaled_df)
X_pca
```

2.	Determine the optimal number of principal components using techniques like scree plot or cumulative explained variance.

```python
explained_var = np.cumsum(pca.explained_variance_ratio_)
explained_var
```

```python
# Scree plot
plt.figure(figsize=(8,5))  
plt.plot(range(1, len(explained_var)+1), explained_var, marker='o')
plt.title("Cumulative Explained Variance")
plt.xlabel("Number of Principal Components")
plt.ylabel("Cumulative Explained Variance")
plt.grid(True)
plt.show()
```

```python
#optimal components (e.g:- 95% variance) 
n_components = np.argmax(explained_var >= 0.95) + 1 
print(f"Number of components explaining 95% variance: {n_components}")
```

```python
#3.	Transform the original dataset into the principal components.
pca_final = PCA(n_components=n_components) 
pca_transformed = pca_final.fit_transform(X_scaled_df)
```

```python
X_pca_df = pd.DataFrame(pca_transformed) 
X_pca_df
```

Task 3: Clustering with Original Data:
1.	Apply a clustering algorithm (e.g., K-means) to the original dataset.


```python
from sklearn.cluster import KMeans 
from sklearn.metrics import silhouette_score, davies_bouldin_score
```

```python
kmeans_org = KMeans(n_clusters = 3, random_state = 42)
labels_org = kmeans_org.fit_predict(X_scaled_df)
```

2.	Visualize the clustering results using appropriate plots.

```python
plt.figure(figsize = (7,5))
plt.scatter(X_scaled_df.iloc[:, 0], X_scaled_df.iloc[:, 2], c=labels_org, cmap = 'viridis')
plt.title("K-mean clusters (original data)")
plt.xlabel(X_scaled_df.columns[0])
plt.ylabel(X_scaled_df.columns[2])
plt.show()
```

3.	Evaluate the clustering performance using metrics such as silhouette score or Davies–Bouldin index.

```python
silhouette_org = silhouette_score(X_scaled_df, labels_org)
db_org = davies_bouldin_score(X_scaled_df, labels_org)
print(f"\nOriginal Data -> Silhouette Score: {silhouette_org:.3f}, Davies–Bouldin Index: {db_org:.2f}")
```

Task 4: Clustering with PCA Data:
1.	Apply the same clustering algorithm to the PCA-transformed dataset.


```python
kmeans_pca = KMeans(n_clusters = 3, random_state=42)    
labels_pca = kmeans_pca.fit_predict(X_pca_df)
```

```python
plt.figure(figsize=(7,5))
plt.scatter(X_pca_df.iloc[:,0],X_pca_df.iloc[:,1],c=labels_pca,cmap='viridis')
plt.title("K-Means Clusters (PCA Data)")
plt.xlabel("PC1")
plt.ylabel("PC2")
plt.show()
```

```python
silhouette_pca = silhouette_score(X_pca_df, labels_pca)
db_pca = davies_bouldin_score(X_pca_df, labels_pca)
print(f"PCA Data -> Silhouette Score: {silhouette_pca:.3f}, Davies–Bouldin Index: {db_pca:.2f}")
```

Task 5: Comparison and Analysis:
1.	Compare the clustering results obtained from the original dataset and PCA-transformed data.


```python
comparison_df = pd.DataFrame({
    "Dataset": ["Original Data", "PCA Data"],
    "Silhouette Score": [silhouette_org, silhouette_pca],
    "Davies–Bouldin Index": [db_orig, db_pca]
})
print("\nComparison of Clustering Performance:\n")
print(comparison_df)
```
